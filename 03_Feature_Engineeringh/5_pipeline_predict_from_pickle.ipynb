# $Without$ $Pipeline$

```python
import pickle
import numpy as np
import warnings
warnings.filterwarnings('ignore', category=UserWarning)


# Load models
ohe_sex = pickle.load(open('models/ohe_sex.pkl','rb'))
ohe_embarked = pickle.load(open('models/ohe_embarked.pkl','rb'))
clf = pickle.load(open('models/clf.pkl','rb'))

# User input
# Pclass / Sex / Age / SibSp / Parch / Fare / Embarked
test_input = np.array(
    [2, 'male', 31.0, 0, 0, 10.5, 'S'],
    dtype=object
).reshape(1, 7)

# One-hot encoding (convert sparse → dense)
test_input_sex = ohe_sex.transform(test_input[:, 1].reshape(1, 1)).toarray()
test_input_embarked = ohe_embarked.transform(test_input[:, -1].reshape(1, 1)).toarray()

# Age
test_input_age = test_input[:, 2].reshape(1, 1)

# Remaining numerical features
test_input_rem = test_input[:, [0, 3, 4, 5]].astype(float)

# Final feature vector
test_input_transformed = np.concatenate(
    (test_input_rem, test_input_age, test_input_sex, test_input_embarked),
    axis=1
)

# Prediction
clf.predict(test_input_transformed)
```

# $With$ $Pipeline$

```python
import pickle
import pandas as pd

pipe = pickle.load(open('pipe.pkl','rb'))

# user input as DataFrame
test_input_df = pd.DataFrame(
    [[2, 'male', 31.0, 0, 0, 10.5, 'S']],
    columns=['Pclass', 'Sex', 'Age', 'SibSp', 'Parch', 'Fare', 'Embarked']
)

pipe.predict(test_input_df)
```
